```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:\\Users\\aktiw\\Downloads\\student_performance_300.csv")
print("Dataset shape:", df.shape)
df.head()
```

```python
print("Data types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
print("Descriptive statistics:")
display(df.describe().round(2))

print("\nAverage Marks:", round(df["Marks"].mean(),2))
print("Average Attendance:", round(df["Attendance"].mean(),2))
print("Average Study Hours:", round(df["Study_Hours"].mean(),2))
```

```python
print("Correlation matrix:")
display(df[["Study_Hours","Attendance","Assignment_Score","Marks"]].corr().round(2))
```

```python
plt.figure(figsize=(9,5))
plt.hist(df["Marks"], bins=15)
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.figure(figsize=(9,5))
plt.scatter(df["Study_Hours"], df["Marks"], alpha=0.65)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
plt.figure(figsize=(9,5))
plt.scatter(df["Attendance"], df["Marks"], alpha=0.65)
plt.title("Attendance vs Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")
plt.show()
```

```python
plt.figure(figsize=(9,5))
df.groupby("Age")["Marks"].mean().plot(kind="bar")
plt.title("Average Marks by Age")
plt.xlabel("Age")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(7,4))
df["Result"].value_counts().plot(kind="bar")
plt.title("Pass/Fail Distribution")
plt.xlabel("Result")
plt.ylabel("Students")
plt.show()
```
